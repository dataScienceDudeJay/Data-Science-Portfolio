```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

plt.style.use("ggplot")
sns.set_theme(style="whitegrid")
```

```python
orders = pd.read_csv("C:/Users/jsero/OneDrive/Desktop/Analyst Portfolio/Global Sales Enterprise/processed_orders.csv")
customer_segments = pd.read_csv("C:/Users/jsero/OneDrive/Desktop/Analyst Portfolio/Global Sales Enterprise/customer_segments.csv")
executive_kpis = pd.read_csv("C:/Users/jsero/OneDrive/Desktop/Analyst Portfolio/Global Sales Enterprise/executive_kpis.csv")

```

________________________________________________________

Executive KPI Layer 

1. KPI TABLE

```python
total_revenue = orders["sales"].sum()
total_profit = orders["profit"].sum()
total_customers = (orders["customer_id"].nunique())
profit_margin = (total_profit / total_revenue)
executive_summary = pd.DataFrame({"total_revenue": [total_revenue],"total_profit": [total_profit], "profit_margin": [profit_margin],"total_customers": [total_customers]})
executive_summary
```

_______________________________

Sales Dashboard Tables 

1. Monthly Sales Table

```python
orders["order_date"] = pd.to_datetime(orders["order_date"])

monthly_sales = (orders.groupby(orders["order_date"].dt.to_period("M")).agg({"sales": "sum","profit": "sum","order_id": "count"}).reset_index())
monthly_sales.columns = ["month","total_sales","total_profit","total_orders"]
```

2. Regional Sales Table

```python
regional_sales = (orders.groupby("region").agg({"sales": "sum","profit": "sum"}).reset_index())
regional_sales
```

________________________________________

1. Customer Segment Summary

```python
segment_summary = (customer_segments.groupby("segment_name").agg({"monetary": "sum","frequency": "mean","customer_id": "count"}).reset_index())
segment_summary.columns = ["segment_name","total_revenue","avg_frequency","customer_count"]
segment_summary
```

2. Churn Dashboard Table

```python
churn_summary = (customer_segments.groupby("segment_name")["churn_risk"].mean().reset_index())
churn_summary.columns = ["segment_name","avg_churn_risk"]
churn_summary
```

____________________________________

Product Dashboard Tables

1. Category Performance

```python
category_performance = (orders.groupby("category").agg({"sales": "sum","profit": "sum"}).reset_index())
category_performance
```

2. Product Performance

```python
top_products = (orders.groupby("product_name")["sales"].sum().sort_values(ascending=False).head(20).reset_index())
top_products.head()
```

______________________________________

Geographic Dashboard Tables

1. Country Sales Table

```python
country_sales = (orders.groupby("country").agg({"sales": "sum","profit": "sum","customer_id": "nunique"}).reset_index())
country_sales.columns = ["country","total_sales","total_profit","total_customers"]
```
