Interview Preparation Questions

Technical Questions:

Why did you use K-Means clustering?
To identify customer groups based on purchasing behavior using unsupervised machine learning.

Why did you scale features before clustering?
Feature scaling prevents large-value variables from dominating clustering behavior.

Why did you use RFM analysis?
RFM captures customer activity, engagement, and monetary value effectively.

Why did you use rolling averages?
Rolling averages smooth volatility and help identify long-term trends.

Why did you build Power BI dashboards?
To create interactive executive reporting and business intelligence systems.
